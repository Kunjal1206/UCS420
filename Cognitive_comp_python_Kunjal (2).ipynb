```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

if is_prime(num):
    print("Prime")
else:
    print("Non-Prime")
```

```python
def add_odd(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total = total + i

    return total

num = int(input("Enter the value of n: "))
print("Sum of odd numbers =", add_odd(num))
```

```python
def sum_prime(n):
    sum = 0

    for num in range(2, n + 1):
        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break

        if prime:
            sum = sum + num

    print("Sum of prime numbers:", sum)

num = int(input("Enter the value of n: "))
sum_prime(num)
```

```python
# 7 Math library

import math as m

print("exp(3)      --> ", m.exp(3))        # Exponential
print("log(64,2)   --> ", m.log(64,2))     # Logarithm base 2
print("log10(1000) --> ", m.log10(1000))   # Logarithm base 10
print("m.cos(60)   --> ", m.cos(60))      # Cosine
print("m.sin(90)   --> ", m.sin(90))      # Sine
print("m.tan(45)   --> ", m.tan(45))      # Tangent
print("m.sqrt(625) --> ", m.sqrt(625))    # Square root
print("m.ceil(45.2) --> ", m.ceil(45.2))  # Ceiling
print("m.floor(45.8) --> ", m.floor(45.8)) # Floor
```

```python
# 8.1 Indexing in string

text = "Python Programming"

print("Original string:", text)

# Accessing individual characters
print("First character:", text[0])

# Extracting characters from index 2 to 7
print("Substring:", text[2:8])

# Extracting the string except the last 4 characters
print("Without last 4:", text[:-4])
```

```python
# 8.8 Reverse a String

text = "Programming"

print("Original string :", text)

print("Using [::-1] :", text[::-1])
print("Using [::-2] :", text[::-2])

reversed_text = text[::-1]

print("Reversed string :", reversed_text)
```
