```python
# Q1. List

# Roll Number: 1024160087

# i. Create list L by multiplying each digit by 10
roll_no = 1024160087

L = [int(digit) * 10 for digit in str(roll_no)]

print("Initial List L:", L)


# ii. Add two numbers using append() and insert()

# Using append()
L.append(50)
# 50 is added at the end of the list
print("After append(50):", L)

# Using insert()
L.insert(2, 30)
# 30 is inserted at index 2
print("After insert(2, 30):", L)


# iii. Remove two elements using remove() and pop()

# Using remove()
L.remove(0)
# remove(0) removes the first occurrence of 0
print("After remove(0):", L)

# Using pop()
L.pop()
# pop() removes the last element from the list
print("After pop():", L)


# iv. Sort in ascending and descending order

L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)


# v. Slicing

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])


# vi. List comprehension to find elements greater than average

average = sum(L) / len(L)

print("Average:", average)

new_list = [x for x in L if x > average]

print("Elements greater than average:", new_list)
```

```python
# Q2. Tuple

# First 8 values from the original list L of Q1
scores = (10, 0, 20, 40, 10, 60, 0, 0)

print("Scores tuple:", scores)


# i. Find highest score and its index
highest = max(scores)
highest_index = scores.index(highest)

# Find lowest score and how many times it appears
lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)

print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)


# ii. Reverse the tuple and return the result as a list

# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(scores[::-1])

print("Reversed tuple as a list:", reversed_scores)


# iii. Ask the user for a score and check whether it is present

user_score = int(input("Enter a score: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")


# iv. Attempt to change an element of the tuple

try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

# Tuple is immutable, so its elements cannot be changed directly,
# unlike a list whose elements can be modified.


# v. Tuple unpacking using * operator

first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# Q3. Random Numbers

random.seed(1024160087)
import random
import math

# Set random seed using roll number
random.seed(1024160087)


# i. Generate a list of 100 random numbers between 100 and 900

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)


# ii. Count and print all odd numbers

odd_numbers = [x for x in numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)

print("Number of odd numbers:", len(odd_numbers))


# iii. Count and print all even numbers

even_numbers = [x for x in numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)

print("Number of even numbers:", len(even_numbers))


# iv. Count and print all prime numbers
# Also build the list of prime numbers using list comprehension

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [x for x in numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)

print("Number of prime numbers:", len(prime_numbers))


# v. Find the number occurring most frequently
# and how many times it occurs

frequency = {}

for x in numbers:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1

most_frequent = max(frequency, key=frequency.get)
maximum_count = frequency[most_frequent]

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", maximum_count)
```

```python
# Q4. Sets

# First 8 digits of roll number: 1, 0, 2, 4, 1, 6, 0, 0

digits = [1, 0, 2, 4, 1, 6, 0, 0]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)


# vi. Union of A and B
union = A.union(B)

print("\nUnion of A and B:", union)


# vii. Intersection of A and B
intersection = A.intersection(B)

print("Intersection of A and B:", intersection)


# viii. Difference of A - B and B - A separately
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present in one set but not the other,
# while symmetric_difference() gives all elements that are not common
# to both sets in a single combined set.


# ix. Symmetric difference
symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)


# x. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))


# xi. Ask the user for X and remove it using discard()

X = int(input("\nEnter a value to remove from set A: "))

A.discard(X)

print("Set A after discard:", A)

# discard() is safer than remove() because it does not raise
# an error if the given value is not present in the set.
```
